```python
#Python program for insertion sort
def insertion_sort(arr):
    for i in range(1,len(arr)):
        key = arr[i] #using i index in arr for key
        j = i-1 # subtracting i - 1 for j value
        while j >= 0 and key < arr[j] : #check key is less than array[j..index] value
            arr[j+1] = arr[j]
            j -=1 # 0-1 : -1
        arr[j+1] = key #-1+1 : 0


#Test function
arr = [45,23,54,12,6,2,3]
#i -1,2,3,4,5,6..
#j - 0,1,2,3,4,5,6....

insertion_sort(arr)
print(arr)
```
